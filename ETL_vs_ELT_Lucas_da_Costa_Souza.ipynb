```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import time

# 1. Extração dos dados
dados_pedidos = pd.DataFrame({
    'id_cliente': [1, 2, 3, 1],
    'produto': ['Notebook', 'Mouse', 'Monitor', 'Mouse'],
    'quantidade': [1, 2, 1, 3],
    'preco_unitario': [3000.00, 100.00, 900.00, 100.00]
})

dados_crm = pd.DataFrame({
    'id_cliente': [1, 2, 3],
    'nome': ['Ana', 'Bruno', 'Carla'],
    'email': ['ana@email.com', 'bruno@email.com', 'carla@email.com']
})

# 2. Processo ETL (Transforma antes de carregar)
dados_pedidos_etl = dados_pedidos.copy()
dados_pedidos_etl['valor_total'] = dados_pedidos_etl['quantidade'] * dados_pedidos_etl['preco_unitario']
dados_etl = pd.merge(dados_pedidos_etl, dados_crm, on='id_cliente')
dados_etl = dados_etl.groupby(['id_cliente', 'nome', 'email']).agg({'valor_total': 'sum'}).reset_index()

print("Resultado ETL:")
print(dados_etl)

# 3. Processo ELT (Transforma depois de carregar)
dados_carregados_pedidos = dados_pedidos.copy()
dados_carregados_crm = dados_crm.copy()
dados_carregados_pedidos['valor_total'] = dados_carregados_pedidos['quantidade'] * dados_carregados_pedidos['preco_unitario']
dados_elt = pd.merge(dados_carregados_pedidos, dados_carregados_crm, on='id_cliente')
dados_elt = dados_elt.groupby(['id_cliente', 'nome', 'email']).agg({'valor_total': 'sum'}).reset_index()

print("\nResultado ELT:")
print(dados_elt)

# 4. Visualização dos resultados
dados_etl = dados_etl.sort_values(by='nome')
dados_elt = dados_elt.sort_values(by='nome')

clientes = dados_etl['nome']
valores_etl = dados_etl['valor_total']
valores_elt = dados_elt['valor_total']

x = np.arange(len(clientes))
largura = 0.35

plt.figure(figsize=(10,5))
plt.bar(x - largura/2, valores_etl, largura, label='ETL', color='skyblue')
plt.bar(x + largura/2, valores_elt, largura, label='ELT', color='orange')
plt.xlabel('Cliente')
plt.ylabel('Valor Total de Compra (R$)')
plt.title('Comparação ETL vs ELT por Cliente')
plt.xticks(x, clientes)
plt.legend()
plt.tight_layout()
plt.show()

# 5. Análise de Performance
start_etl = time.time()
dados_pedidos_etl = dados_pedidos.copy()
dados_pedidos_etl['valor_total'] = dados_pedidos_etl['quantidade'] * dados_pedidos_etl['preco_unitario']
resultado_etl = pd.merge(dados_pedidos_etl, dados_crm, on='id_cliente')
resultado_etl_final = resultado_etl.groupby(['id_cliente', 'nome', 'email']).agg({'valor_total': 'sum'}).reset_index()
tempo_etl = time.time() - start_etl

start_elt = time.time()
dados_brutos_pedidos = dados_pedidos.copy()
dados_brutos_crm = dados_crm.copy()
dados_brutos_pedidos['valor_total'] = dados_brutos_pedidos['quantidade'] * dados_brutos_pedidos['preco_unitario']
resultado_elt = pd.merge(dados_brutos_pedidos, dados_brutos_crm, on='id_cliente')
resultado_elt_final = resultado_elt.groupby(['id_cliente', 'nome', 'email']).agg({'valor_total': 'sum'}).reset_index()
tempo_elt = time.time() - start_elt

print(f"\nTempo ETL: {tempo_etl:.6f} segundos")
print(f"Tempo ELT: {tempo_elt:.6f} segundos")
```
